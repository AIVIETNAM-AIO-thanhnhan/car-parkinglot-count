Đầu ra là bảng đặc trưng để **P4** train model. Chỉnh 2 đường dẫn bên dưới rồi Run All.

```
import sys, os
sys.path.append("../src")          # để import windows.py, features.py, build_features.py
import pandas as pd, numpy as np
from build_features import process_folder

# >>> CHỈNH đường dẫn tới thư mục chứa các cặp (*.jpg, *.xml) cùng tên <<<
FOLDER = "sample"                  # vd: mỗi ảnh 2012-09-12_06_05_16.jpg đi kèm .xml
```

```
df = process_folder(FOLDER, window_size=53)   # 53 = WINDOW_SIZE P2 chốt Ngày 1
print("shape:", df.shape)
print(df["label"].value_counts().sort_index())   # 0=trống, 1=có xe, 2=nền
df.head(3)
```

## Kiểm tra: đủ 395 feature, đúng 4 nhóm
